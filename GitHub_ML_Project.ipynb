```python
import requests
import pandas as pd
```

```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)

print("Status Code:", response.status_code)

data = response.json()

print("Number of repositories:", len(data["items"]))
```

```python
df = pd.DataFrame(data["items"])

print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
required_columns = [
    "name",
    "owner",
    "language",
    "stargazers_count",
    "forks_count",
    "watchers_count",
    "open_issues_count",
    "created_at",
    "updated_at",
    "license"
]

df = df[required_columns]

df.head()
```

```python
df["owner"] = df["owner"].apply(
    lambda x: x.get("login") if isinstance(x, dict) else None
)

df["license"] = df["license"].apply(
    lambda x: x.get("spdx_id") if isinstance(x, dict) else None
)

df.head()
```

```python
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("Unknown")
df["owner"] = df["owner"].fillna("Unknown")

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset Shape after removing duplicates:", df.shape)
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])

print(df[["created_at", "updated_at"]].dtypes)
```

```python
df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})

df.head()
```

```python
print("Final Dataset Shape:", df.shape)

print("\nFinal Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df.head(10)
```

```python
df.to_csv("github_projects.csv", index=False)

print("Dataset saved successfully as github_projects.csv")
```

```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("github_projects.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
conn = sqlite3.connect("github_projects.db")

df.to_sql("Repositories", conn, if_exists="replace", index=False)

print("SQLite database created successfully!")
```

```python
query = """
SELECT *
FROM Repositories
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
WHERE stars > 10000;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
WHERE name LIKE '%Machine%';
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, stars, forks, language
FROM Repositories
WHERE stars > 10000
AND forks > 1000;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, stars, language
FROM Repositories
WHERE NOT language = 'Python';
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT name, owner, stars, language
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""

top10 = pd.read_sql_query(query, conn)

top10
```

```python
query = """
SELECT COUNT(*) AS total_repositories
FROM Repositories;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT AVG(stars) AS average_stars
FROM Repositories;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    COUNT(*) AS total_repositories,
    AVG(stars) AS average_stars
FROM Repositories;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    language,
    COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC;
"""

language_counts = pd.read_sql_query(query, conn)

language_counts
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top10["name"], top10["stars"])

plt.xlabel("Number of Stars")
plt.ylabel("Repository")
plt.title("Top 10 Most Popular Machine Learning Repositories")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
